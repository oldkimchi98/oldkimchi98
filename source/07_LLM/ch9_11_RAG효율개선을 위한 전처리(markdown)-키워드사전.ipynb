# 1. knowledge base 구성을 위한 데이터 생성

```python
%%time
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
loader = Docx2txtLoader("tax_docx/소득세법_with_markdown.docx")
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1500,   # 문서를 쪼갤 때 1500글자씩 chunking
    chunk_overlap=20, # 20문자씩 overlap 하여 chunking
    # separators=["\n\n", "\n", " ", ""]
)
# Recursive(재귀적으로) 다음의 순서대로 chunking 시도 : "\n\n", "\n", " ", ""
documents = loader.load_and_split(text_splitter=text_splitter)
len(documents)
```

```python
documents[43]
```

```python
from dotenv import load_dotenv
from langchain_upstage import UpstageEmbeddings
load_dotenv(override=True) # 커널을 재시작하지 않고 .env 덮어쓰기
embedding = UpstageEmbeddings(
    model='solar-embedding-1-large',  # 4096차원, query/passage 자동 구분
    embed_batch_size=10,
    max_retries=6
)
```

```python
dimension = len(embedding.embed_query("소득세법"))
assert dimension == 4096, f"임베딩 차원이 다릅니다: {dimension}"
dimension
```

```python
# pinecone의 index 생성하기
import time
from pinecone import Pinecone, ServerlessSpec
# 0. pinecone 객체 생성
pc = Pinecone() # api_key=os.getenv('PINECONE_API_KEY')

# 1. 인덱스 설정
index_name = 'tax-index-markdown-4096'
namespace = 'tax-markdown-solar-v1'
dimension = len(embedding.embed_query("차원 확인용"))
assert dimension == 4096, f"임베딩 차원이 다릅니다: {dimension}"

# 2. 인덱스 생성(해당 인덱스가 없을 경우)
existing_indexes = [index.name for index in pc.list_indexes()]
# print(index_name in existing_indexes)
if index_name not in existing_indexes:
    pc.create_index(
        name=index_name, 
        dimension=dimension,
        metric='cosine',  # 유사도 측정 방식
        spec=ServerlessSpec(
            cloud='aws',
            region='us-east-1'
        )        
    )
    # 인덱스가 준비될 때까지 대기
    while not pc.describe_index(index_name).status['ready']:
        time.sleep(1) # 1초 대기
    print(f'인덱스 {index_name}를 생성하였습니다')
else:
    print(f'인덱스 {index_name}이 이미 존재합니다')
# 기존 인덱스가 있어도 임베딩과 차원이 같은지 확인
index_info = pc.describe_index(index_name)
if index_info.dimension != dimension:
    raise ValueError(f'인덱스 {index_info.dimension}차원 / 임베딩 {dimension}차원')
```

```python
# 3. 인덱스 정보 확인
print([index.name for index in pc.list_indexes()])
print(pc.describe_index(index_name).status['ready'])
index = pc.Index(index_name)
print(index.describe_index_stats())
```

```python
%%time
# 4. 문서를 임베딩해서 클라우드 vector DB(index)에 저장
from pinecone import Pinecone
from langchain_pinecone import PineconeVectorStore
import hashlib
pc = Pinecone()

# 처음 한 번: True, 업로드한 벡터DB를 load할 때: False
upload_documents = True
if upload_documents:
    # 같은 문서 목록을 재실행할 때 중복 저장되지 않도록 ID 지정
    ids = [hashlib.sha256(f'{i}:{doc.page_content}'.encode('utf-8')).hexdigest()
           for i, doc in enumerate(documents)]
    database = PineconeVectorStore.from_documents(
        documents=documents,
        embedding=embedding,
        index_name=index_name,
        namespace=namespace,
        ids=ids,
        batch_size=10,
        embeddings_chunk_size=10
    )
else:
    # 업로드한 벡터DB를 load할 때
    database = PineconeVectorStore(
        embedding=embedding,
        index_name=index_name,
        namespace=namespace
    )
```

## Langchain

```python
# from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_upstage import ChatUpstage, UpstageEmbeddings
from pinecone import Pinecone
from langchain_pinecone import PineconeVectorStore
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from dotenv import load_dotenv
# 1. 임베딩 & LLM 초기화
load_dotenv()
# embedding = OpenAIEmbeddings(model = "text-embedding-3-large")
# llm = ChatOpenAI(model='gpt-4.1-mini')
embedding = UpstageEmbeddings(model='solar-embedding-1-large', embed_batch_size=10, max_retries=6)
llm = ChatUpstage(model='solar-pro4')

# 2. vector DB load
# pc = Pinecone()
index_name = 'tax-index-markdown-4096'
namespace = 'tax-markdown-solar-v1'
vectorStore = PineconeVectorStore(
    embedding=embedding,
    index_name=index_name,
    namespace=namespace
)

# 3. Retriever 생성 - search_type='similarity' : 가장 유사한 top-k개 추출
retriever = vectorStore.as_retriever(search_type='similarity', 
                                     search_kwargs={'k':2}) # 유사도가 높은 top-2추출

# 4. 프롬프트템플릿
template = f"""당신은 최고의 한국 소득세법 전문가입니다. 다음 문맥을 참고하여 질문에 답하세요. 답을 모르면 모른다고 
답하세요. 최대 3문장으로 간결하게 답변하세요
질문 : {{query}}
문맥 : {{context}}
답변 :"""
prompt = ChatPromptTemplate.from_template(template)
# 5. 검색된 문서를 텍스트로 변환하는 함수
def format_documents(retrieved_docs):
    return "\n--\n".join([doc.page_content.replace('\n\n', ' ') for doc in retrieved_docs])
```

```python
# 6. RAG 렝체인 구성 (LCEL방식)
# 6. RAG 체인 구성
rag_chain = (
    {
        "context":retriever | format_documents,
        "query":RunnablePassthrough() # 질문만 그대로 전달
    }
    | prompt
    | llm
    | StrOutputParser()
)
result = rag_chain.invoke('연봉 5천만원인 직장인의 소득세는 얼마인가요?')
print(result)
```

# 4. 키워드 사전 활용

```python
query = '연봉 5천만원인 직장인의 소득세는 얼마인가요?'
query = '종합소득이 5천만원인 거주자의 소득세는 얼마인가요?'
```

```python
import numpy as np
def cosine_similarity(vec1, vec2):
        '두 벡터의 코사인 유사도를 반환'
        dot_product = np.dot(vec1, vec2)
        norm_vec1 = np.linalg.norm(vec1)
        norm_vec2 = np.linalg.norm(vec2)
        if norm_vec1==0 or norm_vec2==0:
            return 0
        return dot_product / (norm_vec1*norm_vec2)
vec1 = embedding.embed_query("직장인")
vec2 = embedding.embed_query("거주자")
cosine_similarity(vec1, vec2)
```

```python
vec1 = embedding.embed_query("연봉")
vec2 = embedding.embed_query("종합소득")
cosine_similarity(vec1, vec2)
```

```python
vec1 = embedding.embed_query("5천")
vec2 = embedding.embed_query("5,000")
cosine_similarity(vec1, vec2)
```

```python
# 키워드 사전 : 일반사용자 단어 -> 소득세법 용어
from langchain_core.prompts import ChatPromptTemplate
dictionary = ["연봉 -> 종합소득", "사람을 나타내는 표현 -> 거주자", "5천->5,000"]
prompt = ChatPromptTemplate.from_template(f"""사용자의 질문에 맞춰, 우리의 사전을 참고해서 사용자 질문을 변경해주세요.
만약 변경할 필요가 없을 경우, 사용자의 질문을 변경하지 안항도 됩니다. 오직 질문만 리턴해주세요.
- 사전 : {dictionary}
- 질문 : {{question}}""")
```

```python
strOutputParser = StrOutputParser()
strOutputParser.invoke(llm.invoke(prompt.invoke("소득이 높은 남자가 있습니다")))
```

```python
# 위를 langchain으로
dictionary_chain = prompt | llm | StrOutputParser()
dictionary_chain.invoke({"question":"연봉이 5천만원인 직장인은?"})
```

```python
new_chain = dictionary_chain | rag_chain
```

```python
new_chain.invoke("연봉이 5천만원인 직장인의 소득세는 얼마에요?")
```
